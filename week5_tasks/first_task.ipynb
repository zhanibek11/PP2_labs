```python
import arythmetics

arr1 = [1,2,4,5]
arr2 = [6,7,8,9]

def sumList(arr):
    sum = 0
    for x in arr:
        sum = arythmetics.add(sum,x)
    return sum

print(arythmetics.test(sumList(arr1), sum(arr1)))
print(arythmetics.test(sumList(arr2), sum(arr2)))
```

```python
import arythmetics

arr1 = [1,23,3,40,5]
arr2 = [10,11,12,5,18]

def mulList(arr):
    work = 1
    for x in arr:
        work = arythmetics.multiply(work,x)
    return work

print(arythmetics.test(mulList(arr1), 13800))
print(arythmetics.test(mulList(arr2), 118800))
```

```python
import arythmetics

arr1 = [1,23,3,40,5]
arr2 = [10,11,12,5,18]
arr3 = []

def remList(arr1,arr2):
    for num1,num2 in zip(arr1,arr2):
        arr3.append(arythmetics.divide(num1,num2))
    return arr3
    

print(arythmetics.test(remList(arr1,arr2), [10,2,4,8,3]))
```
