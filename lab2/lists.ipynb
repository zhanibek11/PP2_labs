List Comprehension 

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
  if "a" in x:
    newlist.append(x)

print(newlist)
```

```python
newlist = [x for x in fruits if x != "apple"]
```

```python
thislist = [100, 50, 65, 82, 23]
thislist.sort()
print(thislist)
```

```python
def myfunc(n):
  return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key = myfunc)
print(thislist)
```
