```python
#1
def square_generator(n):
    for i in range(n):
        yield (i+1)**2

n=int(input())
squares = square_generator(n)
for x in squares:
    print(x)
```

```python
#2
def evens(n):
    for i in range(n):
        if i%2==0:
            yield i

n = int(input())

print(",".join(str(i) for i in evens(n)))
```

```python
#3
def div(n):
    for i in range(n):  
        if i%3== 0 and i%4 == 0:  
            yield i

n = int(input())

for num in div(n):
    print(num)
```

```python
#4
def squares(a, b):
    for i in range(a, b + 1):
        yield i**2

a = int(input())
b = int(input())

for num in squares(a, b):
    print(num)
```
