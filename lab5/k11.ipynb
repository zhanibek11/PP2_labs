```python
import re
```

```python
with open("row.txt") as f:
    data = f.read()
```

```python
#1
a = re.findall("a.*b", data)
print(a)
```

```python
#2
a = re.findall("a.*bb+|abbb+", data)
print(a)
```

```python
#3
a = re.findall("[a-z]_+[a-z]+", data)
print(a)
```

```python
#4
a = re.findall(r"[A-Z][a-z]+", data)
print(a)
```

```python
#5
a = re.findall(r"a+.b", data)
print(a)
```

```python
#6
a=re.sub(r"[., ]",':',data)
print(a)
```

```python
#7
a=re.sub(r"_",'',data)
print(a)
```
