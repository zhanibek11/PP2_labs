```python
#10
def sortByHeight(a):
    people_heights= []
    for x in a:
        if x!= -1:
            people_heights.append(x)

    people_heights.sort()
    index= 0
    for i in range(len(a)):
        if a[i]!= -1:
            a[i]= people_heights[index]
            index += 1
    return a
a = [-1,150,190,170,-1,-1,160,180]
print(sortByHeight(a))
```
